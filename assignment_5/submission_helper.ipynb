```python
import numpy as np
```

```python
alpha_preds = {
    '100': np.load("predictions_100.npy")
    ,'75': np.load("predictions_75.npy")
    ,'50': np.load("predictions_50.npy")
    ,'25': np.load("predictions_25.npy")
}

np.savez('maxim_and_lorenzo.npz', **alpha_preds)
```

```python
for key, value in alpha_preds.items():
    print(f"Shape for {key}: {value.shape}")
```

```python
import matplotlib.pyplot as plt

fig, axs = plt.subplots(2, 2, figsize=(10, 8))
axs = axs.flatten()
for i, (key, value) in enumerate(alpha_preds.items()):
    axs[i].hist(value.flatten(), bins=50, alpha=0.7)
    axs[i].set_title(f'Predictions Histogram for {key}% Spiral Data')
    axs[i].set_xlabel('Predicted Alpha Values')
    axs[i].set_ylabel('Frequency')
plt.tight_layout()
plt.show()
```
