```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
filenames = [
    "mlp.npy", # (10000,)
    "main.npy", # (10000,)
]

fig, ax = plt.subplots(1, len(filenames) + 1, figsize=(15, 5))

baseline = np.load("spirals.npz")["alpha_train"].flatten()
ax[0].hist(baseline, bins=100, alpha=0.5, label="baseline")

for i, filename in enumerate(filenames, 1):
    data = np.load(filename)
    ax[i].hist(data, bins=100, alpha=0.5, label=filename)
    
for i in range(len(filenames) + 1):
    ax[i].set_xlim(-10, 10)
    ax[i].set_ylim(0, 2000)
    ax[i].legend()
plt.show()
```
